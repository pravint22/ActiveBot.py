```python

import asyncio
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

from telethon import TelegramClient, errors
from telethon.sessions import StringSession

from google.colab import userdata

API_ID = int(userdata.get("API_ID"))
API_HASH = userdata.get("API_HASH")
try:
    SESSION = userdata.get("TG_SESSION")  # saved login (after first run)
except Exception:
    SESSION = ""

TZ = ZoneInfo("Asia/Kolkata")

# chats: numeric ID (supergroup IDs look like -1001234567890), @username, or invite link.
# topic5_id: optional, for supergroups with Topics enabled.
JOBS = [
    {
        "name": "hourly-reminder",
        "chats": [-5582846855,-1003675127443,-1004098628368,-1004368240996,-1002673021848,-1003316262094],
        "text": "heyyyyy!",
        "mode": "interval",
        "every_min": 1,
        "max_runs": None,
    },

]

# Safety: pause between sends so Telegram doesn't flag the account.
GAP_SECONDS = 3


client = TelegramClient(StringSession(SESSION), API_ID, API_HASH)
await client.start()  # prompts via input() boxes in Colab if not logged in
me = await client.get_me()
print("Logged in as:", me.first_name, "| @" + (me.username or "no-username"))

if not SESSION:
    print("\nSave this string as a Colab Secret named TG_SESSION so you don't need to log in again.")
    print("KEEP IT PRIVATE: anyone with it has full access to your Telegram account.\n")
    print(client.session.save())

# Cache your chats so numeric supergroup IDs resolve
await client.get_dialogs()

# %% CELL 4 - scheduler engine
def next_daily(hhmm, now):
    h, m = map(int, hhmm.split(":"))
    t = now.replace(hour=h, minute=m, second=0, microsecond=0)
    return t if t > now else t + timedelta(days=1)


def init_job(job):
    now = datetime.now(TZ)
    job["runs"], job["done"] = 0, False
    if job["mode"] == "interval":
        job["next"] = now + timedelta(minutes=job["every_min"])
    elif job["mode"] == "daily":
        job["next"] = next_daily(job["at"], now)
    else:
        job["next"] = datetime.strptime(job["at"], "%Y-%m-%d %H:%M").replace(tzinfo=TZ)
        if job["next"] <= now:
            print(f"Skipping '{job['name']}': time already passed.")
            job["done"] = True


async def send(chat, text, topic_id=None):
    for _ in range(3):
        try:
            await client.send_message(chat, text, reply_to=topic_id)
            return True
        except errors.FloodWaitError as e:
            print(f"Flood wait {e.seconds}s")
            await asyncio.sleep(e.seconds + 1)
        except (errors.ChatWriteForbiddenError, errors.UserBannedInChannelError) as e:
            print(f"Cannot write to {chat}: {type(e).__name__}")
            return False
        except Exception as e:
            print(f"Error sending to {chat}: {type(e).__name__}: {e}")
            return False
    return False


async def fire(job):
    for chat in job["chats"]:
        ok = await send(chat, job["text"], job.get("topic_id"))
        print(f"[{datetime.now(TZ):%Y-%m-%d %H:%M:%S}] {job['name']} -> {chat}: {'sent' if ok else 'FAILED'}")
        await asyncio.sleep(GAP_SECONDS)
    job["runs"] += 1
    if job["mode"] == "interval":
        job["next"] += timedelta(minutes=job["every_min"])
        if job.get("max_runs") and job["runs"] >= job["max_runs"]:
            job["done"] = True
    elif job["mode"] == "daily":
        job["next"] = next_daily(job["at"], datetime.now(TZ))
    else:
        job["done"] = True


async def scheduler():
    for j in JOBS:
        init_job(j)
    print("Scheduler started. Next runs:")
    for j in JOBS:
        if not j["done"]:
            print(f"  {j['name']}: {j['next']:%Y-%m-%d %H:%M:%S %Z}")
    while True:
        now = datetime.now(TZ)
        for j in JOBS:
            if not j["done"] and now >= j["next"]:
                await fire(j)
        if all(j["done"] for j in JOBS):
            print("All jobs finished.")
            return
        await asyncio.sleep(5)


# await send(JOBS[0]["chats"][0], "Test from my account")

task = asyncio.create_task(scheduler())
# Stop later with: task.cancel()
```
